# A strong triangular-hyperfield matroid violating ME3

This notebook presents and verifies a counterexample referred to in the paper
"On the multiple exchange property of matroids over hyperfields" by Nathan
Bowler, Changxin Ding, and Chayim Lowen. The complete example is specified here:
a rank-$6$ Plücker function on $[12]$, with exactly one zero coordinate.
We verify that it defines a **strong** $T_1$-matroid and that it violates ME3.

## 1. The property ME3

Let $\chi:\binom Er\to T_1$ be a strong Grassmann–Plücker function.
For $B_1,B_2\in\binom Er$ and $I_1\subseteq B_1\setminus B_2$, ME3 asserts
$$
\chi(B_1)\chi(B_2)=0
\quad\Longrightarrow\quad
0\in\mathop{\boxplus}_{\substack{I_2\subseteq B_2\setminus B_1\\|I_2|=|I_1|}}
\chi((B_1\setminus I_1)\cup I_2)\chi((B_2\setminus I_2)\cup I_1).
$$
The two $r$-subsets need not be bases. In fact, the hypothesis requires at
least one of them to be a nonbasis. The domain $I_1\subseteq B_1\setminus B_2$
is used throughout.

In the triangular hyperfield, $a\boxplus b=[|a-b|,a+b]$ and $-a=a$, so signs
have no effect. For nonnegative terms $t_1,\ldots,t_m$, write
$S=\sum_i t_i$ and $M=\max_i t_i$. Then
$$\mathop{\boxplus}_{i=1}^m t_i=[\max(0,2M-S),S].$$
Thus the conclusion of ME3 holds exactly when $2M\le S$.

## 3. Verify the strong matroid axioms

For every $U\in\binom E7$ and $J\in\binom E5$, the strong Plücker relation is
$$0\in\mathop{\boxplus}_{e\in U\setminus J}p_{U\setminus\{e\}}p_{J\cup\{e\}}.$$
The program tests the equivalent inequality $2\max t_e\le\sum t_e$.
Terms with $e\in J$ vanish by alternation. When $J\subseteq U$, the remaining
two products are identical, so the relation is automatic and is skipped.
Every other relation is checked, including those with zero products and those
with more than three formal terms. Nonzeroness is checked as well.

There are $\binom{12}{7}\binom{12}{5}=627264$ indexed relations. Of these,
$\binom{12}{7}\binom75=16632$ are automatic; $610632$ are evaluated explicitly.
The table below counts **formal** terms $|U\setminus J|$, retaining zero products.

For this larger rational example we use the checker's exact-integer mode.
If $D$ is the least common multiple of the coordinate denominators, replacing
$p_S$ by $Dp_S$ multiplies every product and every slack by $D^2>0$.
It cannot change the sign of a slack or turn a nonzero slack into zero.
The reported slacks are converted back to the original scale.
The default `arithmetic="fraction"` is also available and evaluates the
same relations directly with the original exact coordinates.


```python
strong = check_strong_t1(E, r, coordinates, arithmetic="integer")
print("Strong T1-matroid:", strong["is_strong"])
print("Total indexed relations:", strong["relations_total"])
print("Automatic relations skipped:", strong["relations_skipped_trivial"])
print("Relations checked explicitly:", strong["relations_checked"])
print("Violations:", strong["violations"])
print("\nFormal terms | Relations | Failures | Minimum slack")
for k, item in strong["by_terms"].items():
    print(f"{k:12} | {item['checked']:9} | {item['failures']:8} | {item['minimum_slack']}")
assert strong["is_strong"], strong["first_failure"]
assert strong["relations_total"] == 627264
assert strong["relations_skipped_trivial"] == 16632
assert strong["relations_checked"] == 610632
assert strong["violations"] == 0
assert {k: v["checked"] for k, v in strong["by_terms"].items()} == {
    3: 138600, 4: 277200, 5: 166320, 6: 27720, 7: 792}
```

## 4. An explicit failure of ME3

Choose
$$B_1=A\cup B,\qquad B_2=G,\qquad I_1=A.$$
Then $I_1\subseteq B_1\setminus B_2$ and $p_{B_1}p_{B_2}=0$.
For $I_2\in\binom G3$, the exchanged product is
$$p_{B\cup I_2}p_{A\cup(G\setminus I_2)}=y_{I_2}x_{G\setminus I_2}.$$
Reindexing by $I=G\setminus I_2$ gives the same twenty products as
$x_Iy_{G\setminus I}$. Their largest value is $x_{123}y_{456}=1$.
The other nineteen sum to
$$\frac34+\frac{17}{2}\varepsilon=\frac{77}{80}<1.$$
Consequently their hypersum is
$$\left[\frac3{80},\frac{157}{80}\right],$$
which does not contain $0$. The following cell computes the exchange table
from the full coordinate dictionary and verifies every assertion.


```python
witness = exchange_witness(E, r, coordinates, A | B, G, A)
print("B1 =", witness["B1"])
print("B2 =", witness["B2"])
print("I1 =", witness["I1"])
print("\nI2          | G minus I2  | y_I2 * x_(G minus I2)")
for row in witness["rows"]:
    I2 = row["I2"]
    complement = tuple(i for i in sorted(G) if i not in I2)
    print(f"{str(I2):11} | {str(complement):11} | "
          f"{row['left_value']} * {row['right_value']} = {row['product']}")
terms = witness["products"]
largest = max(terms)
others = sum(terms) - largest
lo, hi = witness["hypersum_interval"]
print("\nTarget:", witness["target"])
print("Largest exchanged product:", largest)
print("Sum of the other nineteen products:", others)
print(f"Hypersum: [{lo}, {hi}]")
print("ME3 fails:", witness["violates_ME3"])
assert witness["target"] == 0
assert witness["ME3_hypothesis"] and witness["violates_ME3"]
assert len(terms) == 20 and largest == 1 and terms.count(largest) == 1
assert others == Fraction(77, 80)
assert others == Fraction(3, 4) + Fraction(17, 2) * eps
assert (lo, hi) == (Fraction(3, 80), Fraction(157, 80))
assert Counter(terms) == Counter(
    x_value(I) * y_value(G - frozenset(I)) for I in combinations(sorted(G), 3))
```

## 5. Exhaustive ME3 check

The reusable function `check_ME3(E, r, chi)` checks all allowed triples with
zero target product. Pairs with two positive coordinates are omitted before
forming exchange products because the hypothesis is false. This omission
cannot hide a violation of ME3. The report counts these vacuous cases separately
and includes a full first failure witness.


```python
me3 = check_ME3(E, r, coordinates)
print("ME3 holds:", me3["holds"])
print("All allowed triples:", me3["triples_total"])
print("Triples with zero target checked:", me3["triples_checked"])
print("Triples skipped (nonzero target):", me3["triples_skipped_nonzero"])
print("Violations:", me3["violations"])
print("Violations by exchange size:", me3["violations_by_exchange_size"])
assert not me3["holds"]
assert me3["triples_total"] == 8305836
assert me3["triples_checked"] == 17977
assert me3["triples_skipped_nonzero"] == 8287859
assert me3["violations"] == 2
assert me3["violations_by_exchange_size"] == {3: 2}
assert me3["first_failure"]["violates_ME3"]
# The two failures exchange A or B out of the unique nonbasis.
for I1 in (A, B):
    assert exchange_witness(E, r, coordinates, A | B, G, I1)["violates_ME3"]
```

## Conclusion

The explicitly defined function $\chi$ is a rank-$6$ strong $T_1$-matroid on
$[12]$ with $923$ bases. For $B_1=A\cup B$, $B_2=G$, and $I_1=A$,
the target product is zero but the exchange hypersum is
$[3/80,157/80]$. Hence **ME3 does not hold for all strong $T_1$-matroids**.
